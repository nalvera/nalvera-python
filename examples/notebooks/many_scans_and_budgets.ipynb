## Free tasks, personal credits and labs

```python
nalvera.budgets()
```

## What the folder costs, before anything is uploaded

```python
FOLDER = "scans"

nalvera.estimate_run(
    FOLDER,
    model="totalsegmentator_ct",   # required. See: nalvera.models()
)
```

## Run every scan, with a spending cap

Free tasks pay first, the budget you name pays the rest. A run that would pass `max_credits` does not start.

```python
result = nalvera.run(
    FOLDER,
    model="totalsegmentator_ct",   # required. See: nalvera.models()
    budget="personal",             # default: None (free tasks first, then personal, or asks).
                                   # Options: "personal" | "lab:<id>" | "free" | "ask"
    max_credits=10,                # default: None (NALVERA_MAX_CREDITS, else no cap)
)
result
```

## Follow the queue

```python
finished = nalvera.wait(result)
finished
```

## Save every result

```python
nalvera.download(finished, out="results")  # default: "results"
```
