## The models you may run

## What the run costs, before anything is uploaded

```python
SCAN = "ct_scan.nii.gz"

nalvera.estimate_run(
    SCAN,
    model="totalsegmentator_ct",   # required. See: nalvera.models()
)
```

## Run it

This spends credits from the budget "personal"; free tasks are used first. Running the cell again returns the same task and charges nothing.

```python
tasks = nalvera.run(
    SCAN,
    model="totalsegmentator_ct",   # required. See: nalvera.models()
    budget="personal",             # default: None (free tasks first, then personal, or asks).
                                   # Options: "personal" | "lab:<id>" | "free" | "ask"
)
tasks
```

## Wait, then save the result under the scan's original name

```python
finished = nalvera.wait(tasks)
finished
```

```python
paths = nalvera.download(finished, out="results")  # default: "results"
paths
```

## Open the result

```python
import nibabel as nib
import numpy as np

labels = np.asarray(nib.load(paths[0]).dataobj)
np.unique(labels)
```

After a restart of the kernel, `nalvera.tasks()`, `nalvera.wait()` and `nalvera.download()` find the tasks and results again.
